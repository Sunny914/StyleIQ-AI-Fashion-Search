```python
# Identity and duplicate checks

print("========== ROW / IDENTITY ==========")

print("Total rows:", len(df))

print(
    "Unique product IDs:",
    df["Id_Product"].nunique()
)

print(
    "Duplicate product IDs:",
    df["Id_Product"].duplicated().sum()
)

print(
    "Exact duplicate rows:",
    df.duplicated().sum()
)


print("\n========== MISSING VALUES ==========")

missing = df.isnull().sum()

print(missing)


print("\n========== EMPTY / WHITESPACE VALUES ==========")

for column in df.columns:
    empty_count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(f"{column}: {empty_count}")
```

```python
# Price format analysis

discount_price = df["Discount Price (in Rs.)"].astype(str)
original_price = df["Original Price (in Rs.)"].astype(str)

print("========== DISCOUNT PRICE ==========")

print(
    "Contains comma:",
    discount_price.str.contains(",", regex=False).sum()
)

print(
    "Has surrounding whitespace:",
    (discount_price != discount_price.str.strip()).sum()
)

print(
    "Unique values:",
    discount_price.nunique()
)


print("\n========== ORIGINAL PRICE ==========")

print(
    "Contains comma:",
    original_price.str.contains(",", regex=False).sum()
)

print(
    "Has surrounding whitespace:",
    (original_price != original_price.str.strip()).sum()
)

print(
    "Unique values:",
    original_price.nunique()
)
```

```python
# Find prices that contain characters other than digits, commas, or whitespace

invalid_discount_format = discount_price[
    ~discount_price.str.fullmatch(r"[\d,\s]+")
]

invalid_original_format = original_price[
    ~original_price.str.fullmatch(r"[\d,\s]+")
]

print("Invalid discount-price formats:", len(invalid_discount_format))
print("Invalid original-price formats:", len(invalid_original_format))

print("\nExamples:")
print(invalid_discount_format.head(20).to_list())
print(invalid_original_format.head(20).to_list())
```

```python
# Convert prices temporarily for validation only
discount_numeric = (
    df["Discount Price (in Rs.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(int)
)

original_numeric = (
    df["Original Price (in Rs.)"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(int)
)

print("========== PRICE VALUE VALIDATION ==========")

print("Discount price <= 0:",
      (discount_numeric <= 0).sum())

print("Original price <= 0:",
      (original_numeric <= 0).sum())

print("Discount price > Original price:",
      (discount_numeric > original_numeric).sum())

print("Discount price == Original price:",
      (discount_numeric == original_numeric).sum())

print("Discount price < Original price:",
      (discount_numeric < original_numeric).sum())
```

```python
print("\n========== PRICE RANGES ==========")

print("Discount price:")
print("Min:", discount_numeric.min())
print("Max:", discount_numeric.max())

print("\nOriginal price:")
print("Min:", original_numeric.min())
print("Max:", original_numeric.max())
```

```python
suspicious_prices = df.loc[
    discount_numeric > original_numeric,
    [
        "Id_Product",
        "Brand",
        "Description",
        "Discount Price (in Rs.)",
        "Original Price (in Rs.)"
    ]
]

print(suspicious_prices.head(30).to_string(index=False))
```

```python
discount_percentage = (
    (original_numeric - discount_numeric)
    / original_numeric
    * 100
)

print("========== DISCOUNT PERCENTAGE ==========")

print("Minimum:", discount_percentage.min())
print("Maximum:", discount_percentage.max())

print("\nPercentiles:")
print(
    discount_percentage.quantile(
        [0, 0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99, 1.00]
    )
)
```

```python
print(
    "\nNegative discount percentage:",
    (discount_percentage < 0).sum()
)
```
