# 75 — Runtime observability instrumentation (Phase 13.9-B)

In-memory sink bound per HTTP request; serving and pipeline stages emit events, metrics, and spans.

```python
from fastapi.testclient import TestClient

from productiq.api.app import create_app
from productiq.observability.events import ObservabilityEventName
from tests.api.fakes import RecordingSearchService

app = create_app(search_service=RecordingSearchService())
client = TestClient(app)
client.post("/api/v1/search", json={"query": "dress", "top_k": 5}, headers={"X-Request-ID": "nb-75"})
events, metrics, spans = app.state.observability_sink.snapshot()
{
    "events": len(events),
    "metrics": len(metrics),
    "spans": len(spans),
    "api_started": any(e.event_name is ObservabilityEventName.API_REQUEST_STARTED for e in events),
}
```
