# ProductIQ Phase 3 — Step 3.9: Representation Generation Pipeline

Orchestrates existing product-side builders into one `ProductRepresentationBundle`.

```python
from productiq.representation import (
    ProductRepresentationPipeline,
    build_product_representation_bundle,
    product_representation_bundle_to_dict,
)
from tests.database.catalog_fixtures import make_product_record

record = make_product_record(
    material="cotton",
    description="Striped Slim Fit Shirt with Patch Pocket",
)
bundle = build_product_representation_bundle(record)
print("Structured:", bundle.product.model_dump())
print("Text:", bundle.text)
print("Lexical:", bundle.lexical)
print("Semantic:", bundle.semantic)
print("Filtering keys:", sorted(bundle.filtering.model_dump().keys()))
```

```python
first = product_representation_bundle_to_dict(bundle)
second = product_representation_bundle_to_dict(
    ProductRepresentationPipeline.build(record)
)
assert first == second
print("Repeated pipeline generation is deterministic.")
```
