# ProductIQ Phase 4 — Step 4.4: Lexical Retrieval Foundations

Inverted index, posting lists, TF/DF/length — **no BM25 scores** (Phase 4.5).

```python
from productiq.retrieval import (
    LexicalIndexDocument,
    build_inverted_lexical_index,
    retrieve_lexical_candidates,
    tokenize_lexical_text,
)

documents = [
    LexicalIndexDocument(product_id="P1", lexical_text="Nike black running shoes"),
    LexicalIndexDocument(product_id="P2", lexical_text="Adidas black running shoes"),
    LexicalIndexDocument(product_id="P3", lexical_text="Nike white casual shoes"),
]
index = build_inverted_lexical_index(documents)

for term in ("nike", "black", "running", "shoes"):
    postings = index.lookup_term(term)
    print(f"{term!r} → DF={index.document_frequency(term)} postings={postings}")

query = "black Nike shoes"
print("Query tokens:", tokenize_lexical_text(query))
candidates = retrieve_lexical_candidates(index, query)
print("Candidate product_ids (union, not BM25-ranked):", candidates)
```
