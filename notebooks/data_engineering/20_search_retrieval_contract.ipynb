# ProductIQ Phase 4 — Step 4.2: Search & Retrieval Contract

Demonstrates `RetrievalRequest`, `RetrievalCandidate`, `RetrievalResponse`, and a stub `Retriever` (no BM25/vector).

```python
from dataclasses import dataclass

from productiq.representation import build_query_representation
from productiq.retrieval import (
    RetrievalCandidate,
    RetrievalMethod,
    RetrievalRequest,
    RetrievalResponse,
    RetrievalResponseMetadata,
    Retriever,
    retrieval_response_to_dict,
)

query = build_query_representation("nike cotton shirt men")
request = RetrievalRequest(query=query, top_k=200)
print("top_k (retrieval stage, not final UI count):", request.top_k)


@dataclass(frozen=True)
class DemoRetriever:
    method: RetrievalMethod

    def retrieve(self, request: RetrievalRequest) -> RetrievalResponse:
        return RetrievalResponse(
            candidates=(
                RetrievalCandidate(
                    product_id="P-DEMO-1",
                    score=14.82 if self.method is RetrievalMethod.BM25 else 0.87,
                    method=self.method,
                ),
            ),
            metadata=RetrievalResponseMetadata(requested_top_k=request.top_k),
        )


bm25 = DemoRetriever(method=RetrievalMethod.BM25)
assert isinstance(bm25, Retriever)
response = bm25.retrieve(request)
print(retrieval_response_to_dict(response))
```
