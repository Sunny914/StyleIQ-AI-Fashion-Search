# ProductIQ Phase 3 — Step 3.6: Semantic Product Representation

Deterministic labeled semantic documents for future embeddings (text only; no models).

```python
from __future__ import annotations

from pathlib import Path

import pandas as pd

from productiq.representation import (
    build_lexical_text,
    build_product_representation,
    build_semantic_text_from_canonical,
)

PROCESSED_PARQUET_FILENAME = "product_catalog.parquet"


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists() and (candidate / "src").exists():
            return candidate
    msg = "Could not locate ProductIQ project root"
    raise RuntimeError(msg)


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
PARQUET_PATH = PROJECT_ROOT / "resources" / "processed" / PROCESSED_PARQUET_FILENAME
row = pd.read_parquet(PARQUET_PATH, engine="pyarrow").iloc[0].to_dict()
rep = build_product_representation(row)
semantic = build_semantic_text_from_canonical(rep, row)
lexical = build_lexical_text(rep, description=row.get("description"))
print("Semantic:", semantic[:240], "...")
print("Lexical:", lexical[:240], "...")
print("Distinct:", semantic != lexical)
```
