Load Phase 12.5 baseline artifacts, build an experiment definition, validate the shared envelope, and inspect descriptive metric deltas vs the reference variant.

```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src" / "productiq").is_dir():
    ROOT = ROOT.parents[1]
sys.path.insert(0, str(ROOT / "src"))

from productiq.retrieval.evaluation.search_evaluation import (
    build_default_baseline_comparison_experiment_definition,
    run_search_evaluation_experiment,
    search_evaluation_experiment_result_to_dict,
)
```

```python
definition = build_default_baseline_comparison_experiment_definition(ROOT)
definition.model_dump()
```

```python
result = run_search_evaluation_experiment(definition, ROOT)
print("reference:", result.reference_variant_name)
print("envelope benchmark:", result.envelope.benchmark_name, result.envelope.benchmark_version)
print("comparisons:", len(result.candidate_comparisons))
for note in result.provenance_notes:
    print("note:", note)
```

```python
for row in result.candidate_comparisons:
    print(row.comparison_variant, "mrr_delta", row.mrr_delta)
```
