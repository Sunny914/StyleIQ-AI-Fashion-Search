# ProductIQ Phase 3 — Step 3.8: Query–Product Representation Contract

Typed `QueryRepresentation` (constraints + lexical + semantic intents).

```python
from productiq.representation import (
    QueryFilterConstraints,
    build_query_representation,
    query_representation_to_dict,
)
from productiq.representation.query_contract import (
    QUERY_CONSTRAINT_EXCLUDED_PRODUCT_FIELDS,
    QUERY_TO_FILTERING_FACET_FIELDS,
    QUERY_TO_FILTERING_PRICE_BOUND_FIELDS,
)

query = build_query_representation(
    "nike cotton shirt",
    constraints=QueryFilterConstraints(category_gender="Men", material=["cotton"]),
)
print(query_representation_to_dict(query))
print("Facet fields (query ↔ filtering contract):", sorted(QUERY_TO_FILTERING_FACET_FIELDS))
print("Price bound fields:", sorted(QUERY_TO_FILTERING_PRICE_BOUND_FIELDS))
print("Excluded from query constraints:", sorted(QUERY_CONSTRAINT_EXCLUDED_PRODUCT_FIELDS))
```
