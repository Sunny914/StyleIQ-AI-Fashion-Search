# Phase 14A — Deployment configuration

ProductIQ loads **Settings** + **ServingConfig** from environment variables, then `create_app()` builds FastAPI for local, production-like, or production deployment (Docker/cloud in later phases).

See `docs/architecture/deployment-configuration.md`.

````python
from productiq.config.deployment import load_deployment_configuration
from productiq.config.settings import Settings
from productiq.serving.config import ServingConfig

deployment = load_deployment_configuration(
    settings=Settings(_env_file=None),
    serving_config=ServingConfig(),
)
deployment.public_dict()
````

```text
configuration (env)
    ↓
environment (APP_ENV)
    ↓
FastAPI (create_app)
    ↓
future deployment (14B Docker, 14C cloud)
```
