# ProductIQ Phase 4 — Step 4.5: BM25 Scoring

Phase 4.4 inverted index → BM25 relevance (no embeddings / hybrid).

```python
from productiq.retrieval import (
    BM25Config,
    BM25Scorer,
    LexicalIndexDocument,
    bm25_idf,
    bm25_scored_candidates_to_dict,
    build_inverted_lexical_index,
    score_bm25_lexical_query,
)

documents = [
    LexicalIndexDocument(product_id="P1", lexical_text="Nike black running shoes"),
    LexicalIndexDocument(product_id="P2", lexical_text="Nike black shoes"),
    LexicalIndexDocument(product_id="P3", lexical_text="Adidas running shoes"),
    LexicalIndexDocument(product_id="P4", lexical_text="Nike formal black shirt"),
]
index = build_inverted_lexical_index(documents)
scorer = BM25Scorer(index, config=BM25Config())
stats = scorer.corpus_statistics
print("Corpus:", stats.model_dump())

for term in ("nike", "black", "running", "shoes"):
    df = index.document_frequency(term)
    print(term, "DF=", df, "IDF=", bm25_idf(stats.document_count, df))

query = "black Nike running shoes"
for product_id in index.document_ids:
    print(product_id, "len=", index.document_length(product_id), "BM25=", scorer.score_document(("black", "nike", "running", "shoes"), product_id))

ranked = score_bm25_lexical_query(index, query, top_k=3)
print("Top-3 BM25:", bm25_scored_candidates_to_dict(ranked))
print("Phase 4.4 = candidates; Phase 4.5 = BM25 scores above.")
```
