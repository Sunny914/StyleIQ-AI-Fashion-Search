Contracts only: **no embedding model**, **no vector index**, **no semantic search execution**.

```python
from productiq.representation import build_query_representation
from productiq.retrieval import (
    SemanticVector,
    build_retrieval_query_view,
    cosine_similarity,
    semantic_retrieval_candidate,
    semantic_retrieval_text_from_query,
)

query = build_query_representation(
    "black nike shoes",
    semantic_text="Nike athletic footwear for running",
)
view = build_retrieval_query_view(query)
print("Product field (future index docs): semantic_text")
print("Query field: semantic_intent ->", view.semantic_retrieval_text)
print("Aligned text:", semantic_retrieval_text_from_query(query))

query_vec = SemanticVector(values=(1.0, 0.0, 0.0))
doc_vec = SemanticVector(values=(0.866, 0.5, 0.0))  # manual example
print("cosine similarity:", cosine_similarity(query_vec, doc_vec))

candidate = semantic_retrieval_candidate("460946942002", similarity=0.91)
print("Future RetrievalCandidate:", candidate.model_dump())
```
