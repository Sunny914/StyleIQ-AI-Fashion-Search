# ProductIQ Phase 3 — Step 3.10: Representation Quality Validation

Validate Phase 3.9 bundles for consistency and degeneracy (not retrieval relevance).

```python
from productiq.representation import (
    build_product_representation_bundle,
    validate_canonical_record_representation_quality,
    validate_product_representation_bundle,
)
from tests.database.catalog_fixtures import make_product_record

record = make_product_record(material="cotton", description="Summer shirt")
bundle = build_product_representation_bundle(record)
report = validate_product_representation_bundle(bundle, record=record)
print("valid:", report.valid, "issues:", report.issue_count)
print("coverage note: field_coverage metrics describe presence, not accuracy.")
```

```python
broken = bundle.model_copy(update={"text": ""})
bad = validate_product_representation_bundle(broken, record=record)
print("degenerate example valid:", bad.valid)
print([issue.code for issue in bad.issues])
```

```python
from productiq.representation import validate_canonical_records_representation_quality

records = [
    make_product_record(product_type="shirt"),
    make_product_record(product_type=None, material="cotton"),
]
catalog = validate_canonical_records_representation_quality(records)
print("total:", catalog.total_records)
print("product_type coverage:", catalog.field_coverage["product_type"])
```
