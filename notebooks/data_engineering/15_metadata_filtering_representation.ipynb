# ProductIQ Phase 3 — Step 3.7: Metadata & Filtering Representation

Typed `FilteringRepresentation` for future hard filters (not query parsing).

```python
from __future__ import annotations

from pathlib import Path

import pandas as pd

from productiq.representation import (
    build_filtering_representation_from_canonical,
    build_product_representation,
    filtering_representation_to_dict,
)

PROCESSED_PARQUET_FILENAME = "product_catalog.parquet"


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists() and (candidate / "src").exists():
            return candidate
    msg = "Could not locate ProductIQ project root"
    raise RuntimeError(msg)


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
PARQUET_PATH = PROJECT_ROOT / "resources" / "processed" / PROCESSED_PARQUET_FILENAME
rows = pd.read_parquet(PARQUET_PATH, engine="pyarrow").head(5).to_dict(orient="records")
filtering = build_filtering_representation_from_canonical(rows[0])
print(filtering_representation_to_dict(filtering))
```

```python
rep = build_product_representation(rows[0])
assert rep.model_dump() == build_product_representation(rows[0]).model_dump()
print("ProductRepresentation unchanged after filtering build pipeline.")
```
