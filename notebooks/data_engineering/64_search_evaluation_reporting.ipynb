Load the canonical report artifact and inspect identity, metrics, comparisons, reproducibility, and Markdown output.

```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src" / "productiq").is_dir():
    ROOT = ROOT.parents[1]
sys.path.insert(0, str(ROOT / "src"))

from productiq.retrieval.evaluation.search_evaluation.reporting import (
    REPORT_MARKDOWN_DIR,
    run_default_search_evaluation_report,
    render_search_evaluation_report_markdown,
    validate_search_evaluation_report_artifact,
)
from productiq.retrieval.evaluation.search_evaluation.reporting.report_schema import (
    SEARCH_EVALUATION_REPORT_FILENAME,
    SEARCH_EVALUATION_REPORT_MARKDOWN_FILENAME,
)
```

```python
report = run_default_search_evaluation_report(ROOT)
print("report:", report.metadata.report_name, report.metadata.report_version)
print("benchmark:", report.benchmark.benchmark_name, report.benchmark.benchmark_version)
print("source artifacts:", len(report.reproducibility.source_artifacts))
```

```python
for variant in report.variant_summaries:
    print(variant.variant_name, "mrr", next(r.value for r in variant.metrics if r.metric_name == "mrr"))
```
