# ProductIQ Phase 4 — Step 4.3: Query Representation for Retrieval

Projects Phase 3.8 `QueryRepresentation` into retrieval inputs (lexical, semantic, constraints). No BM25 or embeddings.

```python
from productiq.representation import QueryFilterConstraints, build_query_representation
from productiq.retrieval import (
    RETRIEVAL_PRODUCT_QUERY_ALIGNMENT,
    build_retrieval_query_view,
    retrieval_query_view_to_dict,
)

query = build_query_representation(
    "women's black floral dress under ₹3000",
    lexical_text="women black floral dress",
    semantic_text="women black floral dress",
    constraints=QueryFilterConstraints(
        category_gender="Women",
        color=["black"],
        max_discount_price_inr=3000,
    ),
)
view = build_retrieval_query_view(query)
print("Product ↔ query alignment:", RETRIEVAL_PRODUCT_QUERY_ALIGNMENT)
print(retrieval_query_view_to_dict(view))
```
