# ProductIQ Phase 4 — Step 4.7: BM25 Candidate Retrieval

`BM25Retriever` implements the Phase 4.2 `Retriever` protocol.

```python
from pathlib import Path

from productiq.representation import build_query_representation
from productiq.retrieval import (
    LexicalIndexDocument,
    RetrievalRequest,
    build_inverted_lexical_index,
    create_bm25_retriever,
    create_bm25_retriever_from_index_path,
    retrieval_response_to_dict,
)

mini = build_inverted_lexical_index(
    [
        LexicalIndexDocument(product_id="P1", lexical_text="nike black running shoes"),
        LexicalIndexDocument(product_id="P2", lexical_text="nike white shoes"),
    ]
)
retriever = create_bm25_retriever(mini)
request = RetrievalRequest(query=build_query_representation("nike shoes"), top_k=5)
print(retrieval_response_to_dict(retriever.retrieve(request)))

persisted = Path("resources/processed/bm25_lexical_index.pkl")
if persisted.is_file():
    prod = create_bm25_retriever_from_index_path(persisted)
    print("Persisted index top-3:", retrieval_response_to_dict(
        prod.retrieve(RetrievalRequest(query=build_query_representation("nike black shoes"), top_k=3))
    ))
```
