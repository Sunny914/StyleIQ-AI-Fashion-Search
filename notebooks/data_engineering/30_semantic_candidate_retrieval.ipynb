# ProductIQ Phase 4 — Step 4.13: Semantic Candidate Retrieval

`SemanticRetriever` implements the Phase 4.2 `Retriever` protocol using BGE query encoding and the Phase 4.12 pgvector HNSW index.

```python
from productiq.config.settings import get_settings
from productiq.database.engine import create_engine_from_settings
from productiq.representation import build_query_representation
from productiq.retrieval import RetrievalRequest, retrieval_response_to_dict
from productiq.retrieval.embedding_encoder import create_bge_small_en_v15_encoder
from productiq.retrieval.query_for_retrieval import build_retrieval_query_view
from productiq.retrieval.semantic_retriever import create_semantic_retriever_from_engine

query = build_query_representation("black running shoes")
view = build_retrieval_query_view(query)
print("semantic_intent:", view.semantic_retrieval_text)

settings = get_settings()
engine = create_engine_from_settings(settings)
encoder = create_bge_small_en_v15_encoder(device="cpu", allow_cpu_fallback=True)
retriever = create_semantic_retriever_from_engine(engine, encoder)

request = RetrievalRequest(query=query, top_k=5)
response = retriever.retrieve(request)
print(retrieval_response_to_dict(response))
engine.dispose()
```
