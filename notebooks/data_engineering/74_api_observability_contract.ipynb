# 74 — API observability contract (Phase 13.9.1)

Contract-only: events, metrics, spans. No Prometheus/Grafana/OpenTelemetry integration yet.

```python
from datetime import UTC, datetime

from productiq.observability import (
    OBSERVABILITY_CONTRACT_VERSION,
    ObservabilityEvent,
    ObservabilityEventName,
    ObservabilityOutcome,
    build_observability_context,
)
from productiq.observability.metrics import MetricName, MetricObservation

ctx = build_observability_context(operation="search", request_id="notebook-demo")
event = ObservabilityEvent(
    timestamp_utc=datetime.now(tz=UTC),
    event_name=ObservabilityEventName.SEARCH_REQUEST_COMPLETED,
    request_id=ctx.request_id,
    service_name=ctx.service_name,
    operation=ctx.operation,
    outcome=ObservabilityOutcome.SUCCESS,
    duration_ms=0.0,
    metadata={"result_count": 0, "top_k": 10},
)
metric = MetricObservation(
    name=MetricName.SEARCH_REQUESTS_TOTAL,
    value=1.0,
    labels={"operation": "search", "outcome": "success"},
)
{
    "contract_version": OBSERVABILITY_CONTRACT_VERSION,
    "event": event.event_name.value,
    "metric": metric.name.value,
}
```
