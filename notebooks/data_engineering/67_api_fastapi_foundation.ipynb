# Phase 13.2 — FastAPI Application Foundation

Construct the FastAPI app, call health/readiness, and inspect OpenAPI metadata (no retrieval pipelines).

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src" / "productiq").is_dir():
    ROOT = ROOT.parents[1]
sys.path.insert(0, str(ROOT / "src"))

from fastapi.testclient import TestClient

from productiq.api import create_app
from productiq.serving.versioning import PLANNED_ROUTE_HEALTH, PLANNED_ROUTE_READY
```

```python
app = create_app()
app.title, app.version
```

```python
with TestClient(app) as client:
    health = client.get(PLANNED_ROUTE_HEALTH)
    ready = client.get(PLANNED_ROUTE_READY)
health.json(), ready.json()
```

```python
with TestClient(app) as client:
    openapi = client.get("/api/v1/openapi.json")
openapi.json()["info"]
```
