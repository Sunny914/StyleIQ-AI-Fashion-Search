# Phase 12.2 — Search Benchmark Dataset & Judgments

Load the canonical search benchmark, validate integrity, and compare with the legacy lexical adapter.
No search execution or metric computation.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src" / "productiq").is_dir():
    ROOT = ROOT.parents[1]
sys.path.insert(0, str(ROOT / "src"))

from productiq.retrieval.evaluation.schema import LEXICAL_RETRIEVAL_BENCHMARK_FILENAME
from productiq.retrieval.evaluation.search_evaluation import (
    SEARCH_BENCHMARK_FILENAME,
    SearchEvaluationVariant,
    build_search_evaluation_request,
    catalog_provenance_from_benchmark,
    convert_lexical_benchmark_path_to_search,
    default_search_benchmark_path,
    load_search_evaluation_benchmark,
    summarize_search_benchmark_integrity,
    validate_search_benchmark_integrity,
)
```

```python
eval_dir = ROOT / "resources" / "evaluation"
benchmark_path = default_search_benchmark_path(ROOT)
assert benchmark_path.name == SEARCH_BENCHMARK_FILENAME

benchmark = load_search_evaluation_benchmark(benchmark_path)
validate_search_benchmark_integrity(benchmark)
summary = summarize_search_benchmark_integrity(benchmark)
provenance = catalog_provenance_from_benchmark(benchmark)

print("benchmark_name:", benchmark.benchmark_name)
print("benchmark_version:", benchmark.benchmark_version)
print("query_count:", summary.query_count)
print("judgment_count:", summary.judgment_count)
print("grade_distribution:", dict(summary.grade_distribution))
print("catalog_artifact:", provenance.catalog_artifact)
print("checksum:", (provenance.source_representation_checksum or "")[:16], "...")
```

```python
legacy_path = eval_dir / LEXICAL_RETRIEVAL_BENCHMARK_FILENAME
adapted = convert_lexical_benchmark_path_to_search(legacy_path)
validate_search_benchmark_integrity(adapted)
print("legacy adapter identity:", adapted.benchmark_name, adapted.benchmark_version)
print(
    "adapted judgments:",
    summarize_search_benchmark_integrity(adapted).judgment_count,
)
```

```python
variant = SearchEvaluationVariant(
    variant_name="notebook_placeholder",
    variant_version="0.0.0",
    description="Runner not implemented in 12.2.",
)
request = build_search_evaluation_request(benchmark, variant)
print("evaluation_version:", request.evaluation_version)
print("request queries:", len(request.benchmark.queries))
print("min_relevant_grade:", request.metric_configuration.min_relevant_grade)
print("k_values:", request.metric_configuration.k_values)
```
