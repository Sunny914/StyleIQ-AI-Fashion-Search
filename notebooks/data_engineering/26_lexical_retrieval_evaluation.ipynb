# ProductIQ Phase 4 — Step 4.8: Lexical Retrieval Evaluation

Load the manually curated benchmark, run BM25 via the generic `Retriever` evaluator, and inspect aggregate and per-query metrics.

```python
from pathlib import Path

from productiq.retrieval import create_bm25_retriever_from_index_path
from productiq.retrieval.evaluation import (
    LexicalRetrievalEvaluator,
    load_lexical_retrieval_benchmark,
)

benchmark_path = Path("resources/evaluation/lexical_retrieval_benchmark_v1.json")
index_path = Path("resources/processed/bm25_lexical_index.pkl")

benchmark = load_lexical_retrieval_benchmark(benchmark_path)
retriever = create_bm25_retriever_from_index_path(index_path)
evaluator = LexicalRetrievalEvaluator(retrieval_top_k=50, k_values=(1, 5, 10, 20, 50))
result = evaluator.evaluate(benchmark, retriever)

print("Aggregate:", result.aggregate.model_dump())
print()
for row in result.per_query:
    print(row.query_id, row.query_text)
    print("  P@10:", row.metrics.precision_at_k.get(10))
    print("  R@10:", row.metrics.recall_at_k.get(10))
    print("  RR:", row.metrics.reciprocal_rank)
    print("  first relevant rank:", row.metrics.first_relevant_rank)
    if row.metrics.reciprocal_rank == 0.0:
        print("  top retrieved:", row.metrics.retrieved_product_ids[:5])
    print()
```
